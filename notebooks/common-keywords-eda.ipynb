```python
import pandas as pd 

df = pd.read_csv('../data/raw/common-keywords.csv')

df.head()
```

## Cleaning the dataset

### Dropping rows with NULL values

```python
df.dropna(inplace = True)
```

### Removing duplicates

```python
df.drop_duplicates(inplace = True)
```

## Saving the Cleaned Dataset

```python
df.to_csv('../data/processed/common-keywords-processed.csv', index = False)
```

## Exploratory Data Analysis

```python
cleaned_df = pd.read_csv('../data/processed/common-keywords-processed.csv')

cleaned_df.info()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
plt.pie(cleaned_df['Type'].value_counts(), labels=cleaned_df['Type'].value_counts().index)

plt.show
```

Keywords belonging to the type "Technical and Digital" are overrepresented, the other types roughly have the same amount of keywords. Despite the plurality holding significatnly more than the type with the second most number of keywords "General", this doesn't warrant adding more keywords unless absolutely necessary, because this dataset is going to be compared to a another paragraph, having too many keywords will make it too sensitive. 

Additionally, Technical and Digital keywords are overrepresented due to the nature of that industry, for example two job postings for a "Software Engineer" in the same company could have a completely different set of skills, tech stack, etc, often they'll be specific in the requirements section, therefore it is necessary for that industry to have a large number of keywords.

It can be argued that the Healthcare and Medical industry has a wide array of positions with different requirements, however, the title of "General Pratitioner" often times carries the same requirements throughout different hospitals.
